```python
# ============================================================
# CREDIT RISK PREDICTION
# File: inference.py
#
# Purpose:
# - Load the trained model
# - Accept information about a new applicant
# - Make a credit-risk prediction
# - Display prediction probability
# ============================================================


# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import joblib
import pandas as pd


# ============================================================
# 2. LOAD TRAINED MODEL
# ============================================================

MODEL_PATH = "models/best_model.pkl"

model = joblib.load(MODEL_PATH)


# ============================================================
# 3. CREATE NEW APPLICANT DATA
# ============================================================

applicant = {

    "Status_Checking": "A11",

    "Duration": 12,

    "Credit_History": "A32",

    "Purpose": "A43",

    "Credit_Amount": 2500,

    "Savings": "A61",

    "Employment": "A73",

    "Installment_Rate": 2,

    "Personal_Status_Sex": "A93",

    "Other_Debtors": "A101",

    "Residence_Since": 4,

    "Property": "A121",

    "Age": 35,

    "Other_Installment_Plans": "A143",

    "Housing": "A152",

    "Existing_Credits": 1,

    "Job": "A173",

    "Liable_People": 1,

    "Telephone": "A192",

    "Foreign_Worker": "A201"
}


# ============================================================
# 4. CONVERT APPLICANT INTO DATAFRAME
# ============================================================

applicant_df = pd.DataFrame(
    [applicant]
)


# ============================================================
# 5. FEATURE ENGINEERING
# ============================================================

# Credit amount relative to loan duration
applicant_df["Credit_Amount_per_Duration"] = (
    applicant_df["Credit_Amount"]
    /
    applicant_df["Duration"]
)


# Credit amount relative to age
applicant_df["Credit_Amount_per_Age"] = (
    applicant_df["Credit_Amount"]
    /
    applicant_df["Age"]
)


# Loan duration relative to age
applicant_df["Duration_Age_Ratio"] = (
    applicant_df["Duration"]
    /
    applicant_df["Age"]
)


# Credit amount relative to existing credits
applicant_df["Credit_Amount_per_Existing_Credit"] = (
    applicant_df["Credit_Amount"]
    /
    (applicant_df["Existing_Credits"] + 1)
)


# ============================================================
# 6. MAKE PREDICTION
# ============================================================

prediction = model.predict(
    applicant_df
)


probability = model.predict_proba(
    applicant_df
)[0][1]


# ============================================================
# 7. CONVERT PREDICTION INTO BUSINESS LABEL
# ============================================================

if prediction[0] == 1:

    risk_label = "BAD CREDIT RISK"

else:

    risk_label = "GOOD CREDIT RISK"


# ============================================================
# 8. DISPLAY RESULT
# ============================================================

print("\n" + "=" * 60)

print("CREDIT RISK PREDICTION")

print("=" * 60)


print("\nApplicant Information:")
print(applicant_df.to_string(index=False))


print("\nPrediction:")
print(risk_label)


print("\nProbability of Bad Credit Risk:")
print(
    f"{probability * 100:.2f}%"
)


print("\nProbability of Good Credit Risk:")
print(
    f"{(1 - probability) * 100:.2f}%"
)


print("=" * 60)
```
